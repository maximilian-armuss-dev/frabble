# Model playground

Send one prepared scenario transition to a model, validate its answer, and compare the move with the saved witness.

## Choose the input and model

Select a scenario source, then narrow it by name, board size, and sampling round. `TRANSITION_INDEX` chooses the board state inside that scenario. Model names come from `config/model_configs.yaml`. The selection cell only prepares and shows the prompt; it does not call a provider.

```python
from visualization.src import run_figures as figures
from visualization.src import notebook_views as view
```

```python
SCENARIO_SOURCE = "generated/7r_final_merged"
SCENARIO_NAME = None
BOARD_SIZE = 10
SAMPLING_ROUND = 0
TRANSITION_INDEX = 0
MODEL_NAME = "openai_gpt-5"
REASONING_EFFORT = "high"

prepared = view.prepare_selected_model_run(
    SCENARIO_SOURCE, name=SCENARIO_NAME, board_size=BOARD_SIZE,
    sampling_round=SAMPLING_ROUND,
    transition_index=TRANSITION_INDEX,
    model_name=MODEL_NAME,
    reasoning_effort=REASONING_EFFORT,
)
```

## Send the request

The next cell makes a real provider request and may incur cost. Check the selected model and prompt above before running it.

```python
timed_response = figures.call_prepared_llm_transition(prepared)
```

```python
context = figures.finalize_llm_transition(prepared, timed_response)
view.show(figures.display_llm_run_summary(context))
```

## Submitted move

The parsed model answer on the board. Blue tiles are new; green tiles reuse matching board cells; red tiles mark conflicts.

```python
view.show_all(figures.plot_llm_run_move(context, move_source="parsed"))
```

## Compare with the witness

The witness proves that the case is solvable. It is not the only valid move and may not be the highest-scoring move.

```python
view.show_all(figures.plot_llm_run_move(context, move_source="ground_truth"))
```
