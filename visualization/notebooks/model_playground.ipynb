# Model Playground

Choose a prepared puzzle and inspect a model answer. `saved` loads an existing result. `fresh` previews the prompt before a separate, potentially paid model request.

## 1. Choose a puzzle

Run the next cell, then use the dropdowns to choose a case set, dimension, number of sequences on the board, and sampling round.

```python
from visualization.src import notebook_views as view
picker = view.show_case_picker()
```

## 2. Choose mode and model

Edit only the three values below. `saved` loads a previous answer for this puzzle and model. `fresh` prepares a new prompt; `REASONING_EFFORT` applies only in that mode. Then run the short cell that follows.

```python
MODE = "saved"  # "saved" or "fresh"
MODEL_NAME = "or_gpt-5-5"
REASONING_EFFORT = "low"  # used only in "fresh" mode
```

```python
session = view.open_model_playground(picker, MODE, MODEL_NAME, REASONING_EFFORT)
```

## 3. Send only in `fresh` mode

**This cell sends a new model request in `fresh` mode and may cost money.** It sends nothing in `saved` mode. If you change the puzzle after previewing it, run step 2 again first.

```python
await session.send()
```

## 4. Inspect the answer

Run the next cell. In 2D, inspect Scrabble-style tiles and colored multiplier squares; in 3D, rotate the grounded board and select a multiplier plane. A 5D answer is checked and shown as text; board plots currently support up to 4D.


```python
session.show_answer()
```
