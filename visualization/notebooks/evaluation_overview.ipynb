# Evaluation Overview

Explore saved evaluation results across models, board sizes and dimensions. This notebook reads artifacts and makes no model requests.

## 1. Choose the data

Run the next cell. Select a case set and evaluation run, then check the dimensions, sequence counts, and sampling rounds to include. All values are selected initially.

```python
from visualization.src import overview_figures as figures
from visualization.src import notebook_views as view
filters = view.show_overview_filter(default_case_set="pilot")
```

## 2. Load the summary

Run this cell after changing any checkbox or the run. It rebuilds the numbers from matching attempts. Then rerun the charts you want to update.

```python
aggregate = view.load_overview_selection(filters)
view.show(figures.display_run_summary(aggregate))
```

## Valid moves and failure rates

One full-width heatmap per model, with board size across columns and dimension across rows. Use the menu to switch between pass, truncation, semantic-failure and request-error rates. Cells show counts and denominators; gray cells have no measured rate. **Green always means better:** more passes or fewer failures. Hover for outcome and failure details.

Pass, truncation and semantic-failure rates use completed responses; request-error rates use all scheduled attempts. Provider-reported response errors remain in the saved completed-response denominator.

```python
view.show_all(figures.plot_rate_heatmaps(aggregate))
```

## Outcome composition

All scheduled attempts, including valid moves and provider failures. Models with the same combination of failure categories are grouped together. Within each group, models are sorted by their share of valid moves, highest first. Models whose attempts all truncated are placed last. Green denotes valid moves, red truncations, blue semantic failures, and gray provider/request errors. Hover for exact counts and shares.

```python
view.show_all(figures.plot_outcome_composition(aggregate))
```

## Quality of valid moves

Use the menu for score / certified optimum or absolute word score. Each **circle** represents one valid move. The blue gradient becomes darker with increasing dimension. Each model has its own full-width row, with a common vertical scale across models.

Board sizes use their actual numeric spacing. Small horizontal offsets separate dimensions and repeated rounds. Invalid or missing moves are not scored as zero. Hover identifies the case and optimal score; historical cases without a certified optimum appear only in the absolute-score view.

```python
view.show_all(figures.plot_move_quality(aggregate))
```

## Token usage

Each model has a full-width row with numeric board-size spacing. The blue gradient identifies dimension; marker shape identifies outcome. Switch between completion, input and reasoning tokens. Dashed lines in the completion view show configured budgets. Missing usage is omitted rather than treated as zero.

```python
view.show_all(figures.plot_token_usage(aggregate))
```

## Runtime

One full-width row per model, with actual board-size spacing. Duration is in minutes and includes recorded retries and retry waits. The blue gradient identifies dimension and marker shape identifies outcome. Missing durations are omitted.

```python
view.show_all(figures.plot_runtime(aggregate))
```
