# Evaluation results

Inspect pass rates, failure classes, move quality, token use, and runtime for one evaluation run.

## Choose a run

`RUN_CONFIG` is the filename of a run config without `.yaml`. The table lists all completed runs made with that config, newest first. Set `RUN = 1` for the first row, another row number, or paste an exact run ID.

```python
from visualization.src import evaluation_figures as figures
from visualization.src import notebook_views as view
```

```python
RUN_CONFIG = "or_7r_all_but_qwen"
RUN = 1

aggregate = view.select_evaluation_run(RUN_CONFIG, RUN)
```

## Pass rates

The share of completed provider responses that produced a valid move. Transport errors are reported separately and are not part of this denominator.

```python
view.show_all(figures.plot_pass_rate_heatmaps(aggregate))
```

## Primary failure composition

Why attempts failed: for example truncation, an invalid sequence, a spatial conflict, or a missing board overlap.

```python
view.show_all(figures.plot_primary_failure_bars(aggregate))
```

## Success across sampled grammars

Shows whether a result is consistent across the independently generated cases or driven by a few easy samples.

```python
view.show_all(figures.plot_grammar_pass_rates(aggregate))
```

## Move quality on passing attempts

Compares valid model moves with the saved witness. The witness proves feasibility, but it is not necessarily the best possible move.

```python
view.show_all(figures.plot_quality_score_bars(aggregate))
```

## Average token usage

Provider-reported input and completion tokens for the selected run.

```python
view.show_all(figures.plot_token_usage_bars(aggregate))
```

## Average runtime

Observed request duration. This includes provider behavior and network conditions, not only model computation.

```python
view.show_all(figures.plot_latency_tables(aggregate))
```
