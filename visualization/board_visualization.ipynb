# Board visualization

Interactive Plotly views for sparse multidimensional Scrabble boards.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from visualization.board_figures import (
    animate_scenario_2d,
    load_scenario_json,
    plot_board_3d,
    scenario_boards_and_placements,
)
```

```python
SCENARIO_PATH = ROOT / "outputs" / "generator_v1.json"
scenario = load_scenario_json(SCENARIO_PATH)
boards, placements = scenario_boards_and_placements(scenario)

{
    "scenario": scenario["config_name"],
    "steps": len(boards),
    "dimensions": boards[-1].dimensions,
    "final_tiles": len(boards[-1].cells),
}
```

## 3D view
