# Scenario animation

Dimension-agnostic scenario animation. Choose a scenario JSON, visible axes, and optional slices for hidden dimensions.

```python
from visualization.board_figures import (
    animate_scenario_projected,
    default_visible_axes,
    load_scenario_json,
    scenario_boards_and_placements,
)
```

```python
SCENARIO_NAME = "generator_v1"

scenario = load_scenario_json(SCENARIO_NAME)
boards, placements = scenario_boards_and_placements(scenario)

VISIBLE_AXES = default_visible_axes(boards[0].dimensions)
SLICE_COORDS = {}

{
    "scenario": scenario["config_name"],
    "steps": len(boards),
    "dimensions": boards[0].dimensions,
    "visible_axes": VISIBLE_AXES,
    "slice_coords": SLICE_COORDS,
    "final_tiles": len(boards[-1].cells),
}
```

## Construction sequence

```python
animate_scenario_projected(
    scenario,
    visible_axes=VISIBLE_AXES,
    slice_coords=SLICE_COORDS,
    title=scenario["config_name"],
)
```
