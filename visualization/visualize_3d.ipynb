# Board visualization

Interactive Plotly views for sparse multidimensional Scrabble boards.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from visualization.board_figures import (
    animate_scenario_3d,
    load_scenario_json,
    scenario_boards_and_placements,
)
```

```python
SCENARIO_PATH = ROOT / "outputs" / "generator_3d.json"
scenario = load_scenario_json(SCENARIO_PATH)
boards, placements = scenario_boards_and_placements(scenario)

{
    "scenario": scenario["config_name"],
    "steps": len(boards),
    "dimensions": boards[-1].dimensions,
    "final_tiles": len(boards[-1].cells),
}
```

## 3D view

```python
VISIBLE_AXES = (0, 1, 2)
# For dimensions > 3, choose one coordinate on each hidden axis.
SLICE_COORDS = {axis: 0 for axis in range(boards[-1].dimensions) if axis not in VISIBLE_AXES}

animate_scenario_3d(scenario, axes=VISIBLE_AXES, slice_coords=SLICE_COORDS, camera_zoom=6)
```
